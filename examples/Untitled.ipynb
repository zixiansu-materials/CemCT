```python
from cemct import calculate_phase_fractions
from cemct import load_labelled_tiff
```

```python
file_path = (
    "../data/synthetic_labelled_cement_4phase_100.tif"
)

volume = load_labelled_tiff(file_path)

print(volume.shape)
print(volume.dtype)
```

```python
phases = {
    1: "Unreacted particle",
    2: "Reaction product",
    3: "Pore",
    4: "Other phase",
}
```

```python
results = calculate_phase_fractions(
    volume=volume,
    phases=phases,
    voxel_size_um=0.7,
)

display(results)
```
